# Data Visualization with Matplotlib & Seaborn

This notebook creates histograms, box plots, bar charts, scatter plots, and a correlation heatmap using the Iris dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
OUT = Path('../visualizations')
OUT.mkdir(exist_ok=True)
df = sns.load_dataset('iris')
df.head()
```

```python
df.info()
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='sepal_length', bins=15, kde=True)
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig(OUT/'histogram_sepal_length.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='species', y='petal_length')
plt.title('Petal Length by Species')
plt.tight_layout()
plt.savefig(OUT/'boxplot_petal_length.png', dpi=150)
plt.show()
```

```python
mean_petal = df.groupby('species', as_index=False)['petal_length'].mean()
plt.figure(figsize=(8,5))
sns.barplot(data=mean_petal, x='species', y='petal_length', palette='Set2', hue='species', legend=False)
plt.title('Average Petal Length by Species')
plt.tight_layout()
plt.savefig(OUT/'bar_average_petal_length.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='sepal_length', y='petal_length', hue='species', palette='Set2', s=70)
plt.title('Sepal Length vs Petal Length')
plt.tight_layout()
plt.savefig(OUT/'scatter_sepal_vs_petal.png', dpi=150)
plt.show()
```

```python
corr = df.drop(columns=['species']).corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig(OUT/'correlation_heatmap.png', dpi=150)
plt.show()
```
